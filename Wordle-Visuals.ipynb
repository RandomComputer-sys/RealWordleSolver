```python
import pandas as pd
from Wordle_Game import Wordle
game = Wordle()
word = list(game.getWord())
guesses = 1
data = {
    1: [0, 0, 0, 0, 0],
    2: [0, 0, 0, 0, 0],
    3: [0, 0, 0, 0, 0],
    4: [0, 0, 0, 0, 0],
    5: [0, 0, 0, 0, 0]
}
        
while guesses < 6 and ([1, 1, 1, 1, 1,] not in data.values()):
    data[guesses] = game.board(word)


    print(data[guesses])
    guesses += 1
if [1, 1, 1, 1, 1,] not in data.values():
    print("You have run out of Guesses!")
    print(data)
            
else:
    print(f"You completed the Wordle in {guesses -1} trys!")
    print(data)

#df = pd.DataFrame(data)
#df
```
